## Una nuova vita del device
Una cartella nuova (`nuovo`): il device non ricorda il certificato revocato.

```jupytersharp
if (-not $cfg -or -not $out) { throw "Mancano le variabili dello scenario: esegui prima la cella di caricamento (la prima cella di codice) di questo notebook." }
$id = 'device-da-revocare'
$nuovo = Open-ScenarioDevice -Config $cfg -DeviceId $id -Notebook '11-device-nuova-registrazione' -Fresh
Invoke-DeviceTick $nuovo
$s = Get-DeviceCertStatus $nuovo
"certificato: $($s.State) | Hub $($nuovo.AssignedHub)"
$nuovo.Events | ForEach-Object { "  evento: $($_.Event)" }
```

## Cosa ne pensa l'Hub

```jupytersharp
$vecchio = Open-ScenarioDevice -Config $cfg -DeviceId $id -Notebook '09-device-revoca'
Show-HubAcceptance -Label 'certificato nuovo' -Hub $nuovo.AssignedHub -DeviceId $id -StateDir $nuovo.Dir
Show-HubAcceptance -Label 'certificato revocato di prima' -Hub $nuovo.AssignedHub -DeviceId $id -StateDir $vecchio.Dir
```

## Cosa abbiamo visto
- La revoca **non impedisce** al device di ripartire: con il bootstrap valido ottiene un certificato che l'Hub accetta.
- Il certificato revocato resta rifiutato.
- Per fermare il device serve un altro gesto: **disabilitarlo** (notebook 12).
