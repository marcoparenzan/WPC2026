```jupytersharp
# Carica la libreria e legge lo scenario (notebooks\_scenario.psd1): da qui in poi si usa $cfg
. ([scriptblock]::Create([IO.File]::ReadAllText((Get-ChildItem . , .\notebooks , .. -Filter _start.ps1 -ErrorAction SilentlyContinue | Select-Object -First 1).FullName)))
```

```jupytersharp
$id = 'device-da-revocare'; $controllo = 'truck-01'
$dev = Open-ScenarioDevice -Config $cfg -DeviceId $id -Notebook '09-device-revoca'
$ctlDir = (Get-ChildItem (Join-Path $cfg.Out "devices\$controllo") -Directory | Sort-Object Name | Select-Object -Last 1).FullName
```

## Il certificato revocato

```jupytersharp
Show-HubAcceptance -Label "$id, dopo la revoca" -Hub $dev.AssignedHub -DeviceId $id -StateDir $dev.Dir
```

## Un altro device sullo stesso Hub
La revoca riguarda solo il device revocato.

```jupytersharp
Show-HubAcceptance -Label "$controllo, non toccato" -Hub $dev.AssignedHub -DeviceId $controllo -StateDir $ctlDir
```

## Il device in ADR
La revoca invalida i certificati, ma il device resta **abilitato**.

```jupytersharp
"stato in ADR: " + (Get-AdrDeviceState -Name $id -Namespace $cfg.AdrName -ResourceGroup $cfg.Rg)
```
