# 07 · Il device rinnova il certificato
*Blocco 3 · Certificati. Passo di esecuzione.*

Il rinnovo **non è una funzione del servizio**: è il device che, prima della scadenza, rifà la richiesta con una chiave e una CSR nuove, esattamente come la prima volta.
Qui il certificato ha ancora molta vita, quindi il rinnovo si **forza**.

**Prima:** solo l'onboarding (notebook 02). Il device si rifà da zero, come al notebook 03, con un'altra identità: `robot-02`.

```jupytersharp
# Carica la libreria e legge lo scenario (notebooks\_scenario.psd1): da qui in poi si usa $cfg
. ([scriptblock]::Create([IO.File]::ReadAllText((Get-ChildItem . , .\notebooks , .. -Filter _start.ps1 -ErrorAction SilentlyContinue | Select-Object -First 1).FullName)))
```

## Il device, rifatto da zero
Una cartella vuota e una prima registrazione, come al notebook 03. Poi ci sarà qualcosa da rinnovare.

```jupytersharp
$id = 'robot-02'
$dev = Open-ScenarioDevice -Config $cfg -DeviceId $id -Notebook '07-device-rinnovo' -Fresh
Invoke-DeviceTick $dev
$prima = Get-DeviceCertStatus $dev
"certificato di partenza: $($prima.Thumbprint)... scade $($prima.NotAfter)"
```

## Il rinnovo
Una nuova registrazione, con una chiave nuova. Il device tiene il certificato vecchio nella cartella `history`.

```jupytersharp
$riuscito = Invoke-DeviceRegistration $dev 'rinnovo forzato'
"registrazione riuscita: $riuscito"
```

## Il certificato è cambiato
Stessa identità, certificato diverso.

```jupytersharp
$dopo = Get-DeviceCertStatus $dev
"prima: $($prima.Thumbprint)..."
"dopo : $($dopo.Thumbprint)..."
"cambiato: $($prima.Thumbprint -ne $dopo.Thumbprint)"
```
