# 12 · Fermare il device: disabilitarlo
*Blocco 3 · Certificati. Passo di esecuzione.*

`Set-AdrDeviceEnablement` porta il device in stato **Disabled**. Da quel momento nemmeno un certificato valido basta, e una nuova registrazione a DPS fallisce. Il notebook 13 lo riabilita, così la demo si può ripetere.

```jupytersharp
# Carica la libreria e legge lo scenario (notebooks\_scenario.psd1): da qui in poi si usa $cfg
. ([scriptblock]::Create([IO.File]::ReadAllText((Get-ChildItem . , .\notebooks , .. -Filter _start.ps1 -ErrorAction SilentlyContinue | Select-Object -First 1).FullName)))
```

## Disabilito il device

```jupytersharp
if (-not $cfg -or -not $out) { throw "Mancano le variabili dello scenario: esegui prima la cella di caricamento (la prima cella di codice) di questo notebook." }
$id = 'device-da-revocare'
"stato in ADR: " + (Set-AdrDeviceEnablement -Name $id -Namespace $cfg.AdrName -ResourceGroup $cfg.Rg -State Disabled)
```

## Il device prova a registrarsi
DPS rifiuta (errore 409000). `az` scrive l'errore su stderr: lo nascondiamo con `2>$null`, altrimenti il kernel ferma la cella.

```jupytersharp
$dis = Open-ScenarioDevice -Config $cfg -DeviceId $id -Notebook '12-device-disabilita' -Fresh
Invoke-DeviceTick $dis 2>$null
"certificato ottenuto: $((Get-DeviceCertStatus $dis).State)     (none = DPS ha rifiutato)"
$dis.Events | ForEach-Object { "  evento: $($_.Event)" }
```
